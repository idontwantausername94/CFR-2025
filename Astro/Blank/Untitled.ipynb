Magnitude of Stars and Galaxies 

```python
import pandas as pd
data_galaxies = pd.read_csv("../Data/galaxyzoo_research.csv")
data_stars = pd.read_csv('../Data/gaia_stars.csv')

def magnitude(u_band, i_band):
    brightness =  u_band - i_band
    return brightness

g_band_stars = data_stars['g band']
r_band_stars = data_stars['r band']
g_band_galaxies = data_galaxies['GMAG']
r_band_galaxies = data_galaxies['RMAG']


print(data_galaxies)
print()
print(data_stars)
```

```python
import matplotlib.pyplot as plt
g_band_stars = data_stars['g band']
r_band_stars = data_stars['r band']
g_band_galaxies = data_galaxies['GMAG']
r_band_galaxies = data_galaxies['RMAG']

plt.hist(g_band_stars, alpha=.3, label="Stars by g Band")

plt.xlabel("Brightness")
print()
plt.hist(r_band_stars, color="green", alpha=.3, label="Stars by r Band")

plt.xlabel("Brightness")
print()
plt.hist(g_band_galaxies, color="red", alpha=.3, label="Galaxy by g Band")

plt.xlabel("Brightness")
print()
plt.hist(r_band_galaxies, color="black", alpha=.4, label="Galaxy by r Band")
plt.title("Galaxy and Star Brightness by g and r Filters")
plt.xlabel("Brightness")
plt.legend()
```
